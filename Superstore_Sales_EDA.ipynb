```python
%%writefile /content/superstore_dashboard/app.py

import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import os

# =========================================================
# PAGE CONFIG
# =========================================================

st.set_page_config(
    page_title="Global Superstore Analytics",
    page_icon="📊",
    layout="wide",
    initial_sidebar_state="expanded"
)

# =========================================================
# PROFESSIONAL DARK THEME
# =========================================================

st.markdown("""
<style>

    .stApp {
        background: #0b1120;
        color: #f8fafc;
    }

    section[data-testid="stSidebar"] {
        background: #111827;
        border-right: 1px solid #263244;
    }

    section[data-testid="stSidebar"] * {
        color: #e5e7eb;
    }

    .main-title {
        font-size: 38px;
        font-weight: 800;
        color: #f8fafc;
        margin-bottom: 4px;
    }

    .subtitle {
        color: #94a3b8;
        font-size: 16px;
        margin-bottom: 25px;
    }

    .kpi {
        background: linear-gradient(145deg, #172033, #111827);
        border: 1px solid #263244;
        border-radius: 16px;
        padding: 20px;
        min-height: 125px;
        box-shadow: 0 8px 25px rgba(0,0,0,0.20);
    }

    .kpi-title {
        color: #94a3b8;
        font-size: 14px;
        font-weight: 600;
        text-transform: uppercase;
        letter-spacing: 0.5px;
    }

    .kpi-value {
        color: #f8fafc;
        font-size: 28px;
        font-weight: 800;
        margin-top: 8px;
    }

    .kpi-sub {
        color: #60a5fa;
        font-size: 13px;
        margin-top: 5px;
    }

    .section-title {
        font-size: 22px;
        font-weight: 750;
        color: #f8fafc;
        margin-top: 30px;
        margin-bottom: 12px;
    }

    .insight-card {
        background: #111827;
        border: 1px solid #263244;
        border-radius: 14px;
        padding: 18px;
        margin-bottom: 12px;
    }

    .insight-card b {
        color: #60a5fa;
    }

    .footer {
        text-align: center;
        color: #64748b;
        padding: 30px 0 10px 0;
        font-size: 13px;
    }

    div[data-testid="stMetric"] {
        background: #111827;
        border: 1px solid #263244;
        padding: 15px;
        border-radius: 14px;
    }

    .stDownloadButton button {
        width: 100%;
    }

</style>
""", unsafe_allow_html=True)


# =========================================================
# DATA LOADING
# =========================================================

CSV_PATH = "/content/superstore_dashboard/superstore_dataset2011-2015.csv"


@st.cache_data
def load_data(path):

    data = pd.read_csv(path)

    # Date conversion
    data["Order Date"] = pd.to_datetime(
        data["Order Date"],
        dayfirst=True,
        errors="coerce"
    )

    data["Ship Date"] = pd.to_datetime(
        data["Ship Date"],
        dayfirst=True,
        errors="coerce"
    )

    # Date features
    data["Order Year"] = data["Order Date"].dt.year
    data["Order Month"] = data["Order Date"].dt.month
    data["Order Month Name"] = data["Order Date"].dt.strftime("%b")
    data["Order Quarter"] = "Q" + data["Order Date"].dt.quarter.astype(str)

    # Shipping duration
    data["Shipping Days"] = (
        data["Ship Date"] - data["Order Date"]
    ).dt.days

    return data


# =========================================================
# LOAD DATA
# =========================================================

if not os.path.exists(CSV_PATH):

    st.error(
        "Dataset not found. Please place "
        "superstore_dataset2011-2015.csv inside "
        "/content/superstore_dashboard/"
    )

    st.stop()


df = load_data(CSV_PATH)


# =========================================================
# SIDEBAR
# =========================================================

with st.sidebar:

    st.markdown(
        "<h2 style='color:#f8fafc;'>📊 Analytics</h2>",
        unsafe_allow_html=True
    )

    st.markdown(
        "<p style='color:#94a3b8;'>Global Superstore Dashboard</p>",
        unsafe_allow_html=True
    )

    st.divider()

    st.markdown("### 🔎 Filters")

    years = sorted(df["Order Year"].dropna().unique())

    selected_years = st.multiselect(
        "Order Year",
        years,
        default=years
    )

    categories = sorted(df["Category"].dropna().unique())

    selected_categories = st.multiselect(
        "Category",
        categories,
        default=categories
    )

    markets = sorted(df["Market"].dropna().unique())

    selected_markets = st.multiselect(
        "Market",
        markets,
        default=markets
    )

    segments = sorted(df["Segment"].dropna().unique())

    selected_segments = st.multiselect(
        "Segment",
        segments,
        default=segments
    )

    st.divider()

    st.markdown("### 📁 Dataset")

    st.write(f"Rows: **{len(df):,}**")
    st.write(f"Columns: **{len(df.columns)}**")

    st.divider()

    st.caption("Global Superstore | EDA Analytics")


# =========================================================
# FILTER DATA
# =========================================================

filtered = df[
    df["Order Year"].isin(selected_years)
    & df["Category"].isin(selected_categories)
    & df["Market"].isin(selected_markets)
    & df["Segment"].isin(selected_segments)
].copy()


# =========================================================
# HEADER
# =========================================================

st.markdown(
    '<div class="main-title">Global Superstore Analytics</div>',
    unsafe_allow_html=True
)

st.markdown(
    '<div class="subtitle">'
    'Interactive Exploratory Data Analysis Dashboard '
    'for Sales, Profit, Customers and Products'
    '</div>',
    unsafe_allow_html=True
)


# =========================================================
# KPI CALCULATIONS
# =========================================================

total_sales = filtered["Sales"].sum()
total_profit = filtered["Profit"].sum()
total_quantity = filtered["Quantity"].sum()
total_orders = filtered["Order ID"].nunique()
total_customers = filtered["Customer ID"].nunique()

profit_margin = (
    total_profit / total_sales * 100
    if total_sales != 0 else 0
)


# =========================================================
# KPI CARDS
# =========================================================

k1, k2, k3, k4, k5 = st.columns(5)

with k1:
    st.markdown(
        f"""
        <div class="kpi">
            <div class="kpi-title">Total Sales</div>
            <div class="kpi-value">${total_sales/1e6:.2f}M</div>
            <div class="kpi-sub">Revenue</div>
        </div>
        """,
        unsafe_allow_html=True
    )

with k2:
    st.markdown(
        f"""
        <div class="kpi">
            <div class="kpi-title">Total Profit</div>
            <div class="kpi-value">${total_profit/1e6:.2f}M</div>
            <div class="kpi-sub">{profit_margin:.2f}% margin</div>
        </div>
        """,
        unsafe_allow_html=True
    )

with k3:
    st.markdown(
        f"""
        <div class="kpi">
            <div class="kpi-title">Orders</div>
            <div class="kpi-value">{total_orders:,}</div>
            <div class="kpi-sub">Unique orders</div>
        </div>
        """,
        unsafe_allow_html=True
    )

with k4:
    st.markdown(
        f"""
        <div class="kpi">
            <div class="kpi-title">Customers</div>
            <div class="kpi-value">{total_customers:,}</div>
            <div class="kpi-sub">Unique customers</div>
        </div>
        """,
        unsafe_allow_html=True
    )

with k5:
    st.markdown(
        f"""
        <div class="kpi">
            <div class="kpi-title">Quantity</div>
            <div class="kpi-value">{total_quantity:,}</div>
            <div class="kpi-sub">Units sold</div>
        </div>
        """,
        unsafe_allow_html=True
    )


# =========================================================
# CHART THEME
# =========================================================

def chart_layout(fig, height=420):

    fig.update_layout(
        template="plotly_dark",
        height=height,
        paper_bgcolor="#111827",
        plot_bgcolor="#111827",
        font=dict(color="#e5e7eb"),
        margin=dict(l=20, r=20, t=55, b=20),
        legend=dict(
            bgcolor="rgba(0,0,0,0)"
        )
    )

    return fig


# =========================================================
# SALES & PROFIT TREND
# =========================================================

st.markdown(
    '<div class="section-title">📈 Sales & Profit Trend</div>',
    unsafe_allow_html=True
)

yearly = (
    filtered
    .groupby("Order Year")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=yearly["Order Year"],
        y=yearly["Sales"],
        mode="lines+markers",
        name="Sales",
        line=dict(width=3)
    )
)

fig.add_trace(
    go.Scatter(
        x=yearly["Order Year"],
        y=yearly["Profit"],
        mode="lines+markers",
        name="Profit",
        line=dict(width=3)
    )
)

fig.update_layout(
    yaxis_title="Amount ($)",
    xaxis_title="Year"
)

st.plotly_chart(
    chart_layout(fig),
    use_container_width=True
)


# =========================================================
# CATEGORY + SEGMENT
# =========================================================

c1, c2 = st.columns(2)

with c1:

    st.markdown(
        '<div class="section-title">📦 Category Performance</div>',
        unsafe_allow_html=True
    )

    category_data = (
        filtered
        .groupby("Category")
        .agg(
            Sales=("Sales", "sum"),
            Profit=("Profit", "sum")
        )
        .reset_index()
        .sort_values("Sales", ascending=False)
    )

    fig = px.bar(
        category_data,
        x="Category",
        y="Sales",
        text_auto=".2s",
        title="Sales by Category"
    )

    st.plotly_chart(
        chart_layout(fig, 380),
        use_container_width=True
    )


with c2:

    st.markdown(
        '<div class="section-title">👥 Segment Sales Mix</div>',
        unsafe_allow_html=True
    )

    segment_data = (
        filtered
        .groupby("Segment")["Sales"]
        .sum()
        .reset_index()
    )

    fig = px.pie(
        segment_data,
        names="Segment",
        values="Sales",
        hole=0.55
    )

    st.plotly_chart(
        chart_layout(fig, 380),
        use_container_width=True
    )


# =========================================================
# MARKET + REGION
# =========================================================

c1, c2 = st.columns(2)

with c1:

    st.markdown(
        '<div class="section-title">🌍 Market Performance</div>',
        unsafe_allow_html=True
    )

    market_data = (
        filtered
        .groupby("Market")
        .agg(
            Sales=("Sales", "sum"),
            Profit=("Profit", "sum")
        )
        .reset_index()
        .sort_values("Sales", ascending=False)
    )

    fig = px.bar(
        market_data,
        x="Market",
        y="Sales",
        color="Profit",
        text_auto=".2s",
        title="Sales by Market"
    )

    st.plotly_chart(
        chart_layout(fig, 400),
        use_container_width=True
    )


with c2:

    st.markdown(
        '<div class="section-title">🗺️ Regional Performance</div>',
        unsafe_allow_html=True
    )

    region_data = (
        filtered
        .groupby("Region")
        .agg(
            Sales=("Sales", "sum"),
            Profit=("Profit", "sum")
        )
        .reset_index()
        .sort_values("Sales", ascending=True)
    )

    fig = px.bar(
        region_data,
        x="Sales",
        y="Region",
        orientation="h",
        color="Profit",
        title="Sales by Region"
    )

    st.plotly_chart(
        chart_layout(fig, 400),
        use_container_width=True
    )


# =========================================================
# TOP PRODUCTS
# =========================================================

st.markdown(
    '<div class="section-title">🏆 Top Products</div>',
    unsafe_allow_html=True
)

product_data = (
    filtered
    .groupby("Product Name")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
)

top_products = (
    product_data
    .nlargest(10, "Sales")
    .sort_values("Sales")
)

fig = px.bar(
    top_products,
    x="Sales",
    y="Product Name",
    orientation="h",
    color="Profit",
    text_auto=".2s",
    title="Top 10 Products by Sales"
)

st.plotly_chart(
    chart_layout(fig, 500),
    use_container_width=True
)


# =========================================================
# SUBCATEGORY
# =========================================================

st.markdown(
    '<div class="section-title">📊 Sub-Category Analysis</div>',
    unsafe_allow_html=True
)

subcat_data = (
    filtered
    .groupby("Sub-Category")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
    .sort_values("Sales", ascending=False)
)

fig = px.bar(
    subcat_data,
    x="Sub-Category",
    y="Sales",
    color="Profit",
    title="Sales and Profit by Sub-Category"
)

fig.update_xaxes(tickangle=-45)

st.plotly_chart(
    chart_layout(fig, 450),
    use_container_width=True
)


# =========================================================
# CORRELATION
# =========================================================

st.markdown(
    '<div class="section-title">🔗 Correlation Analysis</div>',
    unsafe_allow_html=True
)

numeric_cols = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit",
    "Shipping Cost",
    "Shipping Days"
]

corr = filtered[numeric_cols].corr()

fig = px.imshow(
    corr,
    text_auto=".2f",
    aspect="auto",
    title="Numeric Feature Correlations"
)

st.plotly_chart(
    chart_layout(fig, 500),
    use_container_width=True
)


# =========================================================
# SCATTER ANALYSIS
# =========================================================

c1, c2 = st.columns(2)

with c1:

    st.markdown(
        '<div class="section-title">💰 Sales vs Profit</div>',
        unsafe_allow_html=True
    )

    sample = filtered.sample(
        min(5000, len(filtered)),
        random_state=42
    )

    fig = px.scatter(
        sample,
        x="Sales",
        y="Profit",
        color="Category",
        hover_data=["Product Name", "Country"],
        opacity=0.65
    )

    st.plotly_chart(
        chart_layout(fig, 430),
        use_container_width=True
    )


with c2:

    st.markdown(
        '<div class="section-title">🏷️ Discount vs Profit</div>',
        unsafe_allow_html=True
    )

    fig = px.scatter(
        sample,
        x="Discount",
        y="Profit",
        color="Category",
        hover_data=["Product Name"],
        opacity=0.65
    )

    st.plotly_chart(
        chart_layout(fig, 430),
        use_container_width=True
    )


# =========================================================
# LOSS MAKING PRODUCTS
# =========================================================

st.markdown(
    '<div class="section-title">⚠️ Largest Loss-Making Products</div>',
    unsafe_allow_html=True
)

loss_products = (
    product_data
    .nsmallest(10, "Profit")
    .sort_values("Profit")
)

fig = px.bar(
    loss_products,
    x="Profit",
    y="Product Name",
    orientation="h",
    text_auto=".2s",
    title="Products with the Lowest Profit"
)

st.plotly_chart(
    chart_layout(fig, 450),
    use_container_width=True
)


# =========================================================
# CUSTOMERS
# =========================================================

st.markdown(
    '<div class="section-title">👤 Top Customers</div>',
    unsafe_allow_html=True
)

customer_data = (
    filtered
    .groupby("Customer Name")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
    .nlargest(10, "Sales")
)

st.dataframe(
    customer_data.style.format({
        "Sales": "${:,.2f}",
        "Profit": "${:,.2f}"
    }),
    use_container_width=True,
    hide_index=True
)


# =========================================================
# SHIPPING
# =========================================================

st.markdown(
    '<div class="section-title">🚚 Shipping Analysis</div>',
    unsafe_allow_html=True
)

ship_data = (
    filtered
    .groupby("Ship Mode")
    .agg(
        Orders=("Order ID", "nunique"),
        Avg_Shipping_Days=("Shipping Days", "mean")
    )
    .reset_index()
)

ship_data["Avg_Shipping_Days"] = ship_data[
    "Avg_Shipping_Days"
].round(2)

st.dataframe(
    ship_data,
    use_container_width=True,
    hide_index=True
)


# =========================================================
# EXECUTIVE INSIGHTS
# =========================================================

st.markdown(
    '<div class="section-title">💡 Executive Insights</div>',
    unsafe_allow_html=True
)

if len(filtered) > 0:

    top_category = (
        filtered.groupby("Category")["Sales"]
        .sum()
        .idxmax()
    )

    top_market = (
        filtered.groupby("Market")["Sales"]
        .sum()
        .idxmax()
    )

    top_segment = (
        filtered.groupby("Segment")["Sales"]
        .sum()
        .idxmax()
    )

    top_product = (
        filtered.groupby("Product Name")["Sales"]
        .sum()
        .idxmax()
    )

    discount_profit_corr = filtered[
        ["Discount", "Profit"]
    ].corr().iloc[0, 1]

    insights = [
        f"<b>Category:</b> {top_category} generated the highest sales among the selected data.",
        f"<b>Market:</b> {top_market} generated the highest sales among the selected markets.",
        f"<b>Segment:</b> {top_segment} contributed the largest share of sales.",
        f"<b>Product:</b> {top_product} was the highest-selling product.",
        f"<b>Discount & Profit:</b> The correlation is {discount_profit_corr:.2f}. "
        "Correlation does not establish causation."
    ]

    for insight in insights:

        st.markdown(
            f"""
            <div class="insight-card">
                {insight}
            </div>
            """,
            unsafe_allow_html=True
        )


# =========================================================
# DATA QUALITY
# =========================================================

st.markdown(
    '<div class="section-title">🔍 Data Quality</div>',
    unsafe_allow_html=True
)

missing = filtered.isnull().sum()
missing = missing[missing > 0]

q1, q2, q3 = st.columns(3)

with q1:
    st.metric(
        "Filtered Rows",
        f"{len(filtered):,}"
    )

with q2:
    st.metric(
        "Columns",
        f"{len(filtered.columns)}"
    )

with q3:
    st.metric(
        "Missing-Value Columns",
        f"{len(missing)}"
    )


# =========================================================
# DOWNLOAD
# =========================================================

st.markdown(
    '<div class="section-title">⬇️ Export Data</div>',
    unsafe_allow_html=True
)

csv_download = filtered.to_csv(index=False).encode("utf-8")

st.download_button(
    label="📥 Download Filtered Dataset",
    data=csv_download,
    file_name="filtered_superstore_data.csv",
    mime="text/csv"
)


# =========================================================
# FOOTER
# =========================================================

st.markdown(
    """
    <div class="footer">
        Global Superstore Analytics Dashboard<br>
        Exploratory Data Analysis • Sales • Profit • Customers • Products
    </div>
    """,
    unsafe_allow_html=True
)
```
